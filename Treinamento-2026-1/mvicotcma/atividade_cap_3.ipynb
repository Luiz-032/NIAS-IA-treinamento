Lista de tweets que serao analisados:

```python
tweets = [
    "Wow, what a great day today!! #sunshine",
    "I feel sad about the things going on around us. #covid19",
    "I'm really excited to learn Python with @JovianML #zerotopandas",
    "This is a really nice song. #linkinpark",
    "The Python programming language is useful for data science",
    "Why do bad things happen to me?",
    "Apple announces the release of the new iPhone 12. Fans are excited.",
    "Spent my day with family!! #happy",
    "Check out my blog post on common string operations in Python. #zerotopandas",
    "Freecodecamp has great coding tutorials. #skillup"
]
```

Listas de palavras que avaliarao o sentimento da pessoa no momento da publicacao:

```python
happy_words = ['great', 'excited', 'happy', 'nice', 'wonderful',
               'amazing', 'good', 'best']
sad_words = ['sad', 'bad', 'tragic', 'unhappy', 'worst']
```

Resolucao propriamente dita:

```python
# Os comandos antecedidos por "###" sao dispensaveis para o codigo,
#porem auxiliam na visualizacao do processo que esta sendo feito
is_happy=False
cont_happy=0
cont_sad=0

#separar as palavras e formar uma lista listas, usando .split()
for sentences in tweets:
    words=sentences.split()
    ###print(words)
    
#comando repetitivo para passar por todas as palavras
#o indice "i" corresponde as palavras a suas pontuacoes concatenadas
    for i in words:
#comando que retira pontuacao das palavras
        clean_word=i.strip('#!,.:')
        if clean_word in happy_words:
            is_happy=False
            cont_happy=cont_happy+1
            ###print(clean_word)
        if clean_word in sad_words:
            is_happy=False
            cont_sad=cont_sad+1
            ###print(clean_word)
print(f'Na lista de tweets existem {cont_happy} mensagens felizes, e {cont_sad} mensagens tristes') 
```
