```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os.path as path
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

seed = np.random.seed(0)
tabulation = "   "
```

```python
titanic_datasets = path.join('datasets', 'titanic')

test = pd.read_csv(path.join(titanic_datasets, 'test.csv'))
train = pd.read_csv(path.join(titanic_datasets, 'train.csv'))

def printEx1(dframe: pd.DataFrame, tabulation = "   "):
    for col in dframe.columns:
        print(tabulation + col + tabulation + "tipo: " + str(dframe[col].dtype) )

print("Colunas de Train:")
printEx1(train)
print("")
print("Colnas de Test:")
printEx1(test)

print("")

print("Valores Nulos em Train:")
print(train.isna().sum())
print("")
print("Valores Nulos em Test:")
print(test.isna().sum())


train_clean = train.copy()
test_clean = test.copy()
```

```python
print(train_clean.columns)

test_clean.drop(columns=['Cabin', 'Ticket'], inplace=True, errors='ignore')
train_clean.drop(columns=['Cabin', 'Ticket'], inplace=True, errors='ignore')

def fillnaWith(dframe: pd.DataFrame, col: str, mode='MEAN'):
    value_to_fill = 0
    series_no_na = dframe[col].copy().dropna()
    match mode:
        case 'MEAN':
            value_to_fill = series_no_na.mean()
        case 'MEDIAN':
            value_to_fill = series_no_na.median()
        case 'MODE':
            value_to_fill = series_no_na.mode()[0]
        case _:
            print("Undefined case, will default to Mean")
            value_to_fill = series_no_na.mean()
    dframe[col] = dframe[col].fillna(value_to_fill)



fillnaWith(test_clean, 'Age', mode='MEDIAN')
fillnaWith(test_clean, 'Fare', mode='MEDIAN')
fillnaWith(test_clean, 'Embarked', mode='MODE')

fillnaWith(train_clean, 'Age', mode='MEDIAN')
fillnaWith(train_clean, 'Fare', mode='MEDIAN')
fillnaWith(train_clean, 'Embarked', mode='MODE')

print("NaNs em Test:")
print(test_clean.isna().sum())
print("")
print("NaNs em Train:")
print(train_clean.isna().sum())
```

```python
max_age = train_clean["Age"].max()
bins_age = range(0, int(max_age), int(max_age/5))

train_clean["AgeInterval"] = pd.cut(train_clean['Age'], bins=bins_age)
train_clean["FareIntervals"] = pd.qcut(train_clean['Fare'], q=6)
```
